```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
ida_data = pd.read_csv("ida.csv")
volta_data = pd.read_csv("volta.csv")
p3_data = pd.read_csv("periodo3.csv")
p5_data = pd.read_csv("periodo5.csv")
p2_data = pd.read_csv("periodo2.csv")

ida_data = ida_data.to_numpy()
volta_data = volta_data.to_numpy()
p3_data = p3_data.to_numpy()
p5_data = p5_data.to_numpy()
p2_data = p2_data.to_numpy()

plot1 = np.append(ida_data, volta_data, axis = 0)
plot1 = np.append(plot1,p3_data,axis = 0)
plot1 = np.append(plot1,p5_data,axis = 0)
plot1 = np.append(plot1,p2_data,axis = 0)

```

```python



plot2 = pd.read_fwf('dadosLyapunov/espectroSigma.dat')
#picos_x01.dat é o mapa de x e o valore de b variando
plot2 = plot2.to_numpy()
```

```python
import matplotlib.pyplot as plt

# Assuming plot1 and plot2 are arrays containing your data
# Define your data arrays
#plot1 = ...  # Assuming you have data for the first scatter plot
#plot2 = ...  # Assuming you have data for the second scatter plot

# Create subplots
fig, axs = plt.subplots(3)

# Set title
plt.suptitle('Mapa de Bifurcação - variando parâmetro b com zoom',size = 20)

# Plot first subplot
ax2 = axs[0].twinx()
axs[0].scatter(plot1[:,1],plot1[:,3],s=0.1,color = 'g')
ax2.scatter(plot2[:,0],plot2[:,1],s=0.1,color = 'r')
axs[0].set_ylabel('posição máximos - x',size = 15)
ax2.set_ylabel('primeiro expoente de Lyapunov',size = 15)
axs[0].set_title('Primeiro Expoente de Lyapunov vs Posição Máxima - x  (a)',size = 15)

# Plot second subplot
ax3 = axs[1].twinx()
axs[1].scatter(plot1[:,1],plot1[:,3],s=0.1,color = 'g')
ax3.scatter(plot2[:,0],plot2[:,2],s=0.1,color = 'r')
axs[1].set_ylabel('posição máximos - x',size = 15)
ax3.set_ylabel('segundo expoente de Lyapunov',size = 15)
axs[1].set_title('Segundo Expoente de Lyapunov vs Posição Máxima - x  (b)',size = 15)

# Plot third subplot
ax4 = axs[2].twinx()
axs[2].scatter(plot1[:,1],plot1[:,3],s=0.1,color = 'g')
ax4.scatter(plot2[:,0],plot2[:,3],s=0.1,color = 'r')
axs[2].set_ylabel('posição máximos - x',size = 15)
ax4.set_ylabel('terceiro expoente de Lyapunov',size = 15)
axs[2].set_title('Terceiro Expoente de Lyapunov vs Posição Máxima - x  (c)',size = 15)

# Set figure size
fig.set_figwidth(8)
fig.set_figheight(12)

# Set x-axis label for the last subplot
axs[2].set_xlabel('b',size = 15)

# Show plot
plt.tight_layout()
plt.show()
```
