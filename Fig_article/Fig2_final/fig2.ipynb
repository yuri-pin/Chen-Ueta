```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib as mpl
from matplotlib.patches import Rectangle, ConnectionPatch
from mpl_toolkits.axes_grid1.inset_locator import inset_axes
from mpl_toolkits.axes_grid1.inset_locator import mark_inset
from matplotlib.colors import LinearSegmentedColormap
```

```python
#parameters plot
def plot_params(fontsize=18, tick_labelsize=12, legend_fontsize=18):
    """
    Update the parameters of the plot.

    Returns
    -------
    cmap : string
        The color map used in the colored plots.
    """
    plt.clf()
    plt.rc('font', size=fontsize)
    plt.rc('xtick', labelsize=tick_labelsize)
    plt.rc('ytick', labelsize=tick_labelsize)
    plt.rc('legend', fontsize=legend_fontsize)
    plt.rc('text', usetex=True)
    font = {'family' : 'stix'}
    plt.rc('font', **font)
    plt.rcParams["mathtext.fontset"] = "stix"
    mpl.rcParams['axes.linewidth'] = 1.3 #set the value globally
    plt.rc('xtick.major', size=8, width=1.3)
    plt.rc('ytick.major', size=8, width=1.3)
```

```python
bif_x_indo = pd.read_csv("bif_indo_x.dat", sep=r"\s+", header=None, names=["b", "x_max"])
bif_x_voltando = pd.read_csv("bif_voltando_x.dat", sep=r"\s+", header=None, names=["b", "x_max"])

atrator1 = pd.read_csv("atrator/atrator1.dat", sep=r"\s+", header=None, names=["b", "t", "x","y","z"])
atrator2 = pd.read_csv("atrator/atrator2.dat", sep=r"\s+", header=None, names=["b", "t", "x","y","z"])
```

```python
plot_params(fontsize=28, tick_labelsize=18, legend_fontsize=18)

fig, ax = plt.subplots(2, 1, figsize=(6,7.5))
# ======================================================
# Painel (a)
# ======================================================
#########figura a
ax[0].plot(bif_x_indo['b'],bif_x_indo['x_max'],'.',ms=0.5,color='red')
ax[0].plot(bif_x_voltando['b'],bif_x_voltando['x_max'],'.',ms=0.5,color='blue')
ax[0].set_xlim(2.5,4.5)
ax[0].set_ylim(-10,35)
ax[0].set_yticks([-10, 5, 20, 35])
#ax[0].set_yticks(np.arange(-14,-1,4))
#ax[0].set_xticks(np.arange(0.1,1.01,0.25))
#ax[0].set_xticks([0.1,0.35,0.60,0.85, 1])
# Flecha vermelha: sentido crescente de b
ax[0].annotate(
    '',
    xy=(4.5, 33),       # ponta da flecha
    xytext=(4.3, 33),   # início
    arrowprops=dict(
        arrowstyle='->',
        color='red',
        lw=3
    )
)

# Flecha azul: sentido decrescente de b
ax[0].annotate(
    '',
    xy=(4.26, 29),       # ponta da flecha
    xytext=(4.47, 29),   # início
    arrowprops=dict(
        arrowstyle='->',
        color='blue',
        lw=3
    )
)
ax[0].set_xlabel(r"$b$",fontsize=20)
ax[0].set_ylabel(r"$x_{\rm max}$", color='black',fontsize=20)
ax[0].text(0.9,0.05,r"(a)",transform=ax[0].transAxes,fontsize=22,fontweight="bold")

#########figura b
ax[1].plot(atrator1['x'],atrator1['z'],color='blue')
ax[1].plot(atrator2['x'],atrator2['z'],color='red')
ax[1].set_xlim(-20,20)
ax[1].set_ylim(20,40)
#ax[0].set_yticks(np.arange(-14,-1,4))
#ax[0].set_xticks(np.arange(0.1,1.01,0.25))
#ax[0].set_xticks([0.1,0.35,0.60,0.85, 1])
ax[1].set_xlabel(r"$x$",fontsize=20)
ax[1].set_ylabel(r"$z$", fontsize=20)
ax[1].text(0.9,0.05,r"(b)",transform=ax[1].transAxes,fontsize=22,fontweight="bold")

#######################
#######################
ax[0].tick_params(axis='x', pad=1)
ax[0].tick_params(axis='y', pad=1)
ax[1].tick_params(axis='x', pad=1)
ax[1].tick_params(axis='y', pad=1)

# Ajuste fino
pos0 = ax[0].get_position()
ax[0].set_position([pos0.x0, pos0.y0, pos0.width, pos0.height])
pos1 = ax[1].get_position()
ax[1].set_position([pos1.x0, pos1.y0, pos1.width, pos1.height])
####################################
# SAVE
####################################

plt.tight_layout()
plt.savefig("fig2.jpeg",format="jpeg",dpi=600,bbox_inches="tight")
plt.show()
```
