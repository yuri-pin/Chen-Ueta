```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib as mpl
```

```python
bif_x_indo = pd.read_csv("bif_indo_x.dat", sep=r"\s+", header=None, names=["b", "x_max"])
bif_x_voltando = pd.read_csv("bif_voltando_x.dat", sep=r"\s+", header=None, names=["b", "x_max"])
bif_y_indo = pd.read_csv("bif_indo_y.dat", sep=r"\s+", header=None, names=["b", "y_max"])
bif_y_voltando = pd.read_csv("bif_voltando_y.dat", sep=r"\s+", header=None, names=["b", "y_max"])
bif_z_indo = pd.read_csv("bif_indo_z.dat", sep=r"\s+", header=None, names=["b", "z_max"])
bif_z_voltando = pd.read_csv("bif_voltando_z.dat", sep=r"\s+", header=None, names=["b", "z_max"])
```

```python
bif_x_indo_unico = bif_x_indo.drop_duplicates(subset='b').copy()
bif_y_indo_unico = bif_y_indo.drop_duplicates(subset='b').copy()
bif_z_indo_unico = bif_z_indo.drop_duplicates(subset='b').copy()

bif_x_voltando_unico = bif_x_voltando.drop_duplicates(subset='b').copy()
bif_y_voltando_unico = bif_y_voltando.drop_duplicates(subset='b').copy()
bif_z_voltando_unico = bif_z_voltando.drop_duplicates(subset='b').copy()
```

```python
dfs_condicoes_indo = pd.merge(bif_x_indo_unico[['b','x_max']],bif_y_indo_unico[['b','y_max']],on='b')
dfs_condicoes_indo = pd.merge(dfs_condicoes_indo,bif_z_indo_unico[['b','z_max']],on='b')

dfs_condicoes_voltando = pd.merge(bif_x_voltando_unico[['b','x_max']],bif_y_voltando_unico[['b','y_max']],on='b')
dfs_condicoes_voltando = pd.merge(dfs_condicoes_voltando,bif_z_voltando_unico[['b','z_max']],on='b')
```

```python
dfs_condicoes_indo = dfs_condicoes_indo.iloc[::8].copy()
dfs_condicoes_voltando = dfs_condicoes_voltando.iloc[::8].copy()
```

```python
dfs_condicoes_voltando = dfs_condicoes_voltando.sort_values('b',ascending=True)
```
